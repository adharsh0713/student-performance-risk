# Student Performance Prediction: Early Risk Detection

**Goal:** predict `Fail`/`Withdrawn` vs `Pass`/`Distinction` using information from the first 30 course days. This notebook runs the full pipeline: data download, preprocessing, EDA, model baselines, XGBoost tuning, test evaluation, SHAP, and model export.

**Dataset:** Open University Learning Analytics Dataset (OULAD). Original download page: https://analyse.kmi.open.ac.uk/open_dataset. UCI mirror: https://archive.ics.uci.edu/dataset/349/open+university+learning+analytics+dataset.

**Important:** OULAD has no direct classroom-attendance column, so VLE clicks/active days are engagement proxies. Withdrawal dates and final outcomes are excluded from model inputs to avoid leakage.
